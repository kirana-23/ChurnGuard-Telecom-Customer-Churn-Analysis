CONVERTING THESE CATEGORICAL FEATURES TO NUMERICAL FEATURE USING LABELENCODER()--which makes easier for machine learning models to interpret and analyse

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
cols=["international plan","voice mail plan","churn"]
for col in cols:
    df[col]=le.fit_transform(df[col])
```

FEATURE ENGINEERING--which creates new features with the help of already existing features in the dataset

```python
df.columns
```

```python
df["total duration"]=df["total day minutes"]+df["total eve minutes"]+df["total night minutes"]+df["total intl minutes"]
```

```python
df["total calls"]=df["total day calls"]+df["total eve calls"]+df["total night calls"]+df["total intl calls"]
```

```python
df["total charge"]=df["total day charge"]+df["total eve charge"]+df["total night charge"]+df["total intl charge"]
```

```python
df.shape
```

DROPPING COLUMNS THAT DOES'NT CONTRIBUTE TO THE CHURN ANALYSIS--area and phone number columns are not helpful and since we are having charge columns separately so minutes columns does'nt provide any value

```python
df.drop(['phone number', 'area code'], axis=1, inplace=True)
```

```python
df.drop(['total day minutes','total eve minutes','total night minutes','total intl minutes'],axis=1,inplace=True)
```

```python
df.shape
```

```python
df.head(10)
```

PERFORMING SCALING--this is to ensure that the machine learning models treats all the features equally

```python
from sklearn.preprocessing import MinMaxScaler
ms=MinMaxScaler()
cols=['account length','number vmail messages', 'total day calls','total day charge', 'total eve calls','total eve charge', 
      'total night calls','total night charge', 'total intl calls','total intl charge', 'customer service calls',
      'total duration', 'total calls', 'total charge']
for col in cols:
    df[col]=ms.fit_transform(df[[col]])
```
